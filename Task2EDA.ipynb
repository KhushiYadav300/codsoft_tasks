# Task 2: Exploratory Data Analysis (EDA)

## Objective
The objective of this task is to analyze the Superstore dataset using descriptive statistics and visualizations to identify trends, patterns, and business insights.

```python
import pandas as pd
```

```python
df=pd.read_csv("cleaned_superstore.csv")
```

## Basic Information

```python
print("Shape of data:",df.shape)
```

```python
df.describe()
```

## Business Metrics

```python
print("Sum of Sales",df['Sales'].sum())
print("Sum of Profit",df['Profit'].sum())
print("Average Sales",df['Sales'].mean())
```

```python
import matplotlib.pyplot as plt
```

## Chart 1: Sales by Category

This chart shows the total sales generated by each product category. It helps identify which category contributes the most to the company's overall revenue.

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales.plot(kind='bar')

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.show()
```

## Chart 2: Profit by Category

This chart shows the total profit generated by each product category. It helps identify which category contributes the most to the company's profitability.

```python


category_profit = df.groupby('Category')['Profit'].sum()

plt.figure(figsize=(6,6))

plt.pie(category_profit,
        labels=category_profit.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Profit Distribution by Category")

plt.show()
```

## Chart 3: Monthly Sales Trend

This chart shows the monthly sales trend over time. It helps identify seasonal patterns and changes in sales performance throughout the year.

```python
#changing type to datetime
df["Order Date"]=pd.to_datetime(df["Order Date"])
df["Ship Date"]=pd.to_datetime(df["Ship Date"])

# Create a Month-Year column
df['Month_Year'] = df['Order Date'].dt.to_period('M')

# Calculate monthly sales
monthly_sales = df.groupby('Month_Year')['Sales'].sum()

# Convert PeriodIndex to string for plotting
monthly_sales.index = monthly_sales.index.astype(str)

# Plot line chart
monthly_sales.plot(kind='line', marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Month-Year")
plt.ylabel("Sales")
plt.xticks(rotation=45)

plt.show()
```

## Chart 4: Correlation Heatmap

This heatmap shows the correlation between numerical variables in the dataset. It helps identify whether two variables have a positive, negative, or no linear relationship.

```python
import seaborn as sns

# Select only numerical columns
numerical_data = df.select_dtypes(include='number')

# Calculate correlation matrix
correlation = numerical_data.corr()

# Plot heatmap
sns.heatmap(correlation, annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")
plt.show()
```
